```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
# Load Data
url = "https://raw.githubusercontent.com/martj42/international_results/master/results.csv"

df = pd.read_csv(url)

df.head()
```

```python
# Basic Dataset Information
df.info()
```

```python
# Basic Dataset Numerical Stats
df.describe()
```

```python
df.columns
```

### Basic Data Cleaning

```python
# ------ DATA CLEANING ------
# Rename Columns appropriately
df = df.rename(columns={
    "home_score": "home_goals", 
    "away_score": "away_goals", 
})

# Convert date column to datetime data type
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date")
```

```python
df.info()
```

##### Handle Missing Values

```python
# Missing Values Check
df.isna().sum()
```

```python
# Drop missing values from goals columns (home_goals & away_goals)
df = df.dropna(subset=['home_goals', 'away_goals'])
```

##### Convert Goals(Home/Away) to integer data type

```python
# Convert home_goals & away_goals variables int not float
df['home_goals'] = df['home_goals'].astype(int)
df['away_goals'] = df['away_goals'].astype(int)
```

```python
df.info()
```

```python
df = df.reset_index(drop=True)
```

##### Save Clean Data

```python
output_path = "../data/processed/matches_clean.csv"
df.to_csv(output_path, index=False)

print(f"Cleaned Matches Data Saved to location: {output_path}")
```
