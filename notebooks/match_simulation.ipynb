## Simulating One Match with Poisson

We model goals scored by each team using a Poisson distribution:

- Home goals: $G_{home} \sim \text{Poisson}(\lambda_{home})$
- Away goals: $G_{away} \sim \text{Poisson}(\lambda_{away})$

Where:
- $\lambda_{home}$ = expected goals for home team
- $\lambda_{away}$ = expected goals for away team

These values come from our model:

$\lambda = \exp(\beta_0 + \beta_1 \cdot \text{elo\_diff})$

---

## Simulation Idea

We randomly draw goals from each distribution:

- Higher $\lambda$ → higher chance of scoring more goals
- Lower $\lambda$ → fewer goals expected

This gives us one possible match outcome.

```python
import numpy as np

# Example: take average values from your dataset
lambda_home = 1.76
lambda_away = 1.18

# Simulate goals
home_goals = np.random.poisson(lambda_home)
away_goals = np.random.poisson(lambda_away)

print(f"Simulated Match Result:")
print(f"Home Team {home_goals} - {away_goals} Away Team")
```

```python
n_simulations = 10000

home_goals_sim = np.random.poisson(lambda_home, n_simulations)
away_goals_sim = np.random.poisson(lambda_away, n_simulations)

# Probabilities
home_win = np.mean(home_goals_sim > away_goals_sim)
draw = np.mean(home_goals_sim == away_goals_sim)
away_win = np.mean(home_goals_sim < away_goals_sim)

print(f"Home Win: {home_win:.2%}")
print(f"Draw: {draw:.2%}")
print(f"Away Win: {away_win:.2%}")
```
